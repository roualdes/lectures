```python
class Table():
    def __init__(self, num_legs = 4):
        self.num_legs = num_legs

    def legs(self):
        return self.num_legs

    def foo(self, x):
        return self.num_legs * x
tbl = Table()
tabl = Table(num_legs = 11)
```

```python
tbl.legs()
```

```python
tabl.foo(3)
```
