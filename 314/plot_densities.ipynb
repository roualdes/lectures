```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.arange(6) + 1
fx = np.ones(6) * 1/6
```

```python
plt.scatter(x, fx, color = "orange", label = "uniform density")
plt.scatter(np.arange(5) + 1, np.ones(5) / 5, label = "other")
plt.ylabel("density")
plt.xlabel("support")
plt.legend()
```
