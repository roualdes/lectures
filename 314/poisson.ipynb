```python
import numpy as np
import scipy.stats as st
import matplotlib.pyplot as plt
```

Poisson examples from https://roualdes.us/lecturenotes/poisson

```python
# 1. 
P = st.poisson(3.5) # random variable with rate parameter
```

```python
# a.
P.pmf(2) # density function, probability mass function
```

```python
# b.
1 - np.sum(P.pmf(np.arange(3)))
```

```python
# c.
P = st.poisson(3.5 * 7) # average per week
np.sum(P.pmf(np.arange(6)))
```
