```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as spicy
```

```python
N = 30
t = 2.7
2 * (1 - spicy.t(df = N - 1).cdf(np.abs(t))) # p-value
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/roualdes/data/refs/heads/master/penguins.csv")
```

```python
ndx = ~df["body_mass_g"].isna()
x = df.loc[ndx, "body_mass_g"]
xbar = np.mean(x)
N = np.sum(ndx)
s = np.std(x, ddof = 1)
t = (xbar - 4000)/ (s / np.sqrt(N))
```

```python
2 * (1 - spicy.t(df = N - 1).cdf(np.abs(t)))
```

```python
spicy.ttest_1samp(x, popmean = 4000)
```
