```python
import numpy as np
import scipy.stats as st
import matplotlib.pyplot as plt
```

```python
Binomial = st.binom
```

```python
B = Binomial(5, 0.5) # K, p
```

```python
B.rvs(size = 10) # random variables / random numbers
```

```python
B.pmf(2) # f( x = 2 |K, p) / probability mass function / density function
```

```python
B.pmf(3)
```

```python
x = np.arange(6)
f = B.pmf(x)
f
```

```python
np.sum(f)
```

```python
plt.scatter(x, f)
```
