1.  $$0.005^5 \cdot (1 - 0.005)^{100 - 5}$$

The $(1 - 0.005)^{100-5}$ needs to show up since we also need to
consider what the remaining $95$ computers have done.

1.  $${100 \choose 5}$$

Question 2. is interested in any $5$ computers failing to install. It
doesn’t matter which of the $100$ fail. So this is a combination, and
this combination is counted with the `choose` operation.

1.  $${100 \choose 5}0.005^5 \cdot (1 - 0.005)^{100 - 5}$$

This is just the Binomial($n=100$, $p=0.005$) density function evaluated
at $5$.
