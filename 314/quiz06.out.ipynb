```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def cummean(x):
    N = np.size(x)
    out = np.zeros(N) # np.zeros_like(x)
    s = 0
    for n in range(N):
        s += x[n]
        out[n] = s / (n + 1)
    return out
```

```python
rng = np.random.default_rng()
x = rng.normal(size = 100)
```

```python
plt.plot(np.arange(100), cummean(x))
```
