```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
rng = np.random.default_rng()
```

```python
N = 101
p = 0.75
x = rng.binomial(1, p, size = N) # Bernoulli(p = 0.75)
```

```python
np.mean(x == 1)
```

```python
ndx = np.arange(1, N + 1)
cm = np.cumsum(x == 1) / ndx
plt.plot(ndx, cm)
```
