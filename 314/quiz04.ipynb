```python
import numpy as np
```

```python
def mean(x):
    N = np.size(x) # x.size
    s = 0.0
    for i in range(N):
        s += x[i]
    return s / N
```

```python
rng = np.random.default_rng() # instantiating an RNG class
x = rng.normal(size = 10) # use rng instance to generate random numbers => ndarray
x
```

```python
np.isclose(mean(x), np.mean(x)) # don't == floats/doubles
```
