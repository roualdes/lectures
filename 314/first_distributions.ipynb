```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
rng = np.random.default_rng()
```

```python
N = 50 # number of data
x = rng.integers(1, 7, size = N) # Uniform(1, 6)
```

```python
np.mean(x == 4) # ¿close to 1 / 6?
```

```python
np.mean(x) # ¿close to 3.5?
```

```python
cx = np.cumsum(x == 1) # (x)
ndx = np.arange(np.size(x)) + 1
cm = cx / ndx # cumulative mean
```

```python
plt.plot(ndx, cm) # convergence plot
plt.axhline(1 / 6, color = "black", linestyle = "--") # 3.5
```
