```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
rng = np.random.default_rng()
x = rng.normal(size = 10000)
```

```python
plt.hist(x, density = True, histtype = "step");
```

```python
y = rng.exponential(size = 1000)
plt.hist(y, density = True, histtype = "step");
```

```python
z = rng.poisson(size = 10000)
```

```python
plt.hist(z, density = True, histtype = "step", bins = 9);
```

```python
R = 100
N = 501
rng = np.random.default_rng()
ndx = np.arange(N) + 1
for r in range(R):          
    x = rng.normal(size = N)
    cm = np.cumsum(x) / ndx       
    plt.plot(ndx, cm) 
```

```python
cms = np.zeros(R)
for r in range(R):          
    x = rng.normal(size = N)
    cms[r] = np.mean(x)
```

```python
plt.hist(cms, density = True, histtype = "step");
```
