```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
N = 10000
rng = np.random.default_rng()
x = rng.normal(size = N)
xbar = np.mean(x)
np.var(x)
```

```python
ndx = np.arange(N) + 1
cm = np.cumsum(x) / ndx
plt.plot(ndx, cm)
```

```python
cv = np.cumsum((x - xbar) ** 2) / ndx
plt.plot(ndx, cv)
```

```python
np.mean(x >= 0) # P[X >= 0] = P[X in R^+]
```

```python
cp = np.cumsum(x >= 0) / ndx
plt.plot(ndx, cp)
```
