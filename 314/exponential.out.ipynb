## Inverse CDF

```python
def invCDF(p, lamda = 1):
    return -lamda * np.log(1 - p)
```

```python
X = st.expon(scale = 5)
```

```python
X.ppf(0.3)
```

```python
invCDF(0.3, lamda = 5)
```
