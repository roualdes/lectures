```python
import numpy as np
import scipy.stats as st
from scipy.optimize import minimize
```

```python
B = st.bernoulli(0.75)
np.random.seed(291)
x = B.rvs(size = 101)
```

```python
def ll_bernoulli(theta, data):
    return -np.sum(st.bernoulli(theta).logpmf(data))
```

```python
rng = np.random.default_rng()
init = rng.random()
print(init)
minimize(ll_bernoulli, init, args = (x,), bounds = [(1e-5, 1 - 1e-5), ])
```

```python
def ll_exponential(theta, data):
    return -np.sum(st.expon(scale = theta).logpdf(data))
```

```python
E = st.expon(scale = 5.5)
x = E.rvs(size = 1001)
init = rng.random()
minimize(ll_exponential, init, args = (x), bounds = [(1e-5, np.inf)])
```
