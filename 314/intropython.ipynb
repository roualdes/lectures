# Class

```python
class Table():
    def __init__(self, num_legs = 4):
        self.legs = num_legs

    def num_legs(self):
        return self.legs
```

```python
t = Table()
x = t.num_legs()
```

```python
s = Table(num_legs = 7)
s.num_legs()
```
