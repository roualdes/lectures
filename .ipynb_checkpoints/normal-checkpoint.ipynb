```python
import scipy.stats as st
import numpy as np
import matplotlib.pyplot as plt
```

```python
N = st.norm()
x = np.linspace(-5, 5, 101)
fx = N.pdf(x)
plt.plot(x, fx)
```

```python
fig = plt.figure()
ax = fig.add_subplot(111, projection="3d")
X, Y = np.meshgrid(x, x)
XY = np.dstack((X, Y))
MvN = st.multivariate_normal(np.zeros(2), np.diag(np.ones(2)))
Z = MvN.pdf(XY)
ax.plot_surface(X, Y, Z)
```
